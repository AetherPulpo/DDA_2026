# Ejercicio 4 - Conversión de coeficientes a punto fijo

Este notebook arma **solo la parte de conversión** de los coeficientes del filtro FIR a punto fijo, usando `arrayFixedInt` de la librería de la cátedra (`tool._fixedInt`). El diseño de la ROM (con y sin simetría) y el código Verilog quedan para después.

**Firma real de `arrayFixedInt`** (confirmada con `help()` y probando los atributos del objeto devuelto):

```python
arrayFixedInt(intWidth, fractWidth, N, signedMode='S', roundMode='trunc', saturateMode='saturate')
```

- `intWidth` / `fractWidth` se usan **tal cual la notación S(m.n) del enunciado**, sin restar nada: S(16.15) → `intWidth=16, fractWidth=15`. El ancho total resultante (`.width`) coincide con `intWidth` cuando `fractWidth < intWidth` (la clase calcula internamente cuántos bits quedan para signo/parte entera).
- `N` puede ser directamente la lista/array de valores a convertir (no hace falta crear el array vacío y cargarlo después).
- Cada elemento del array devuelto es un objeto `DeFixedInt`, con estos atributos/métodos útiles (verificados a mano):
  - `.value` -> entero con signo, en unidades de LSB
  - `.intvalue` -> el mismo valor pero como entero sin signo (complemento a 2), útil para sacar el binario
  - `.fValue` -> el valor decimal ya cuantizado (lo que efectivamente representa ese punto fijo)
  - `.rep` -> string tipo `"S(16,15)"`
  - `.__hex__()` -> representación en hexadecimal
  - `.width` -> ancho total en bits
  - `.bit()` sin argumentos no devolvió nada útil (dio `None`); para el binario usamos `intvalue` formateado con `width` bits en su lugar.

## Función auxiliar para mostrar cada coeficiente

Junta `value`, `intvalue` (para el binario), `fValue` y `__hex__()` en un solo print prolijo.


````python
def mostrar_fixed(nombre, valores_originales, arr_fixed):
    for i, c in enumerate(arr_fixed):
        binario = format(c.intvalue, f'0{c.width}b')
        print(
            f"{nombre}[{i}] = {valores_originales[i]:>10.7f}  ->  "
            f"bin = {binario}   hex = {c.__hex__()}   "
            f"cuantizado = {c.fValue:.7f}   (rep {c.rep})"
        )
````

## Coeficientes del filtro (en punto flotante)


````python
h = [-0.0456, -0.1703, 0.0696, 0.3094, 0.4521, 0.3094, 0.0696, -0.1703, -0.0456]
h
````

## Conversión de h[n] a S(16.15)


````python
h_fixed = arrayFixedInt(16, 15, h, signedMode='S', roundMode='round', saturateMode='saturate')

print(f"Ancho total obtenido: {h_fixed[0].width} bits (esperado: 16)")
print()
mostrar_fixed("h", h, h_fixed)
````
